```python
import pandas as pd
from sklearn.ensemble import IsolationForest

data = {
    'Age':[25,27,26,28,29],
    'Income':[50000,52000,51000,53000,54000],
    'Height':[170,168,172,171,169],
    'Weight':[65,67,66,68,69],
    'Score':[80,82,81,83,84]
}
df = pd.DataFrame(data)
model = IsolationForest(contamination=0.05, random_state=42)
df['Outlier'] = model.fit_predict(df)
print(df)
```

```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# Generate sample data
X, y = make_blobs(
    n_samples=120,
    n_features=5,
    centers=5,
    cluster_std=1.5,
    random_state=42
)

df = pd.DataFrame(
    X,
    columns=["Feature1", "Feature2", "Feature3", "Feature4", "Feature5"]
)


kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(df)

centroids = kmeans.cluster_centers_

distances = []

for i in range(len(df)):
    cluster = df.loc[i, "Cluster"]
    point = df.iloc[i, :5].values
    centroid = centroids[cluster]

    distance = np.linalg.norm(point - centroid)
    distances.append(distance)

df["Distance_to_Centroid"] = distances

print(df.head(10))
centroids_df = pd.DataFrame(
    centroids,
    columns=["Feature1", "Feature2", "Feature3", "Feature4", "Feature5"]
)

print("\nCluster Centroids:")
print(centroids_df)
avg_distance = df.groupby("Cluster")["Distance_to_Centroid"].mean()

print("\nAverage Distance per Cluster:")
print(avg_distance)
```
